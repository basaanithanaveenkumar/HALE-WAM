# HaleBlocks Vision Backbone Guide

**A complete reference for all 60 registered vision backbones in the HaleBlocks library.**

---

## What is a Vision Backbone?

A vision backbone is a neural network that converts an input image `[B, 3, H, W]` into rich
feature representations that downstream tasks (detection, segmentation, VLM fusion, etc.) consume.
Good backbones produce:

- **`pooled_output`** `[B, D]` — global image embedding (for classification / cross-attention)
- **`patch_tokens`** `[B, N, D]` — per-patch features (for dense tasks, VLM patch grids)
- **`feature_maps`** `list[B, Cᵢ, Hᵢ, Wᵢ]` — multi-scale spatial features (for detection/FPN)

## Repository Structure

```
HaleBlocks/src/hale_core/nn/vision/
├── base.py          ← BaseVisionBackbone ABC + BackboneOutput dataclass
├── registry.py      ← build_vision_backbone(), list_vision_backbones()
├── cnn/
│   ├── resnet.py         ← ResNet-18/34/50/101/152/200
│   ├── resnet_dcn.py     ← ResNet-50/101 + Deformable Conv v2
│   ├── convnext.py       ← ConvNeXt V2 (tiny → huge)
│   ├── mobile_one.py     ← MobileOne S0–S4 (re-param inference)
│   ├── efficientnet.py   ← EfficientNetV2 S/M/L/XL
│   └── internimage.py    ← InternImage (tiny → huge, DCNv3)
├── vit/
│   ├── vit.py            ← Standard ViT (tiny → huge + 22B)
│   ├── dino.py           ← DINOv2 (small → giant, register tokens)
│   ├── eva.py            ← EVA-02 (small/base/large, RoPE + SwiGLU)
│   └── deit3.py          ← DeiT III (small → huge)
├── swin/
│   └── swinv2.py         ← SwinV2 (tiny → giant, Log-CPB + cosine attn)
└── hybrid/
    ├── coatnet.py        ← CoAtNet (0–7, conv stem + rel-attn stages)
    └── maxvit.py         ← MaxViT (tiny → xlarge, block+grid attention)
```

````python
# ── Quick Start ──────────────────────────────────────────────────────────────
import sys
sys.path.insert(0, '../src')

from hale_core.nn.vision import build_vision_backbone, list_vision_backbones

# List all available backbones
names = list_vision_backbones()
print(f'Total registered backbones: {len(names)}')
print()

# Group by family prefix
from collections import defaultdict
families = defaultdict(list)
for n in names:
    prefix = n.split('_')[0] if not n.startswith('efficientnet') else 'efficientnet'
    if n.startswith('resnet') and 'dcn' in n: prefix = 'resnet_dcn'
    elif n.startswith('resnet'): prefix = 'resnet'
    elif n.startswith('mobile'): prefix = 'mobile_one'
    elif n.startswith('convnext'): prefix = 'convnext'
    elif n.startswith('internimage'): prefix = 'internimage'
    elif n.startswith('vit'): prefix = 'vit'
    elif n.startswith('dino'): prefix = 'dino'
    elif n.startswith('eva'): prefix = 'eva'
    elif n.startswith('deit3'): prefix = 'deit3'
    elif n.startswith('swinv2'): prefix = 'swinv2'
    elif n.startswith('coatnet'): prefix = 'coatnet'
    elif n.startswith('maxvit'): prefix = 'maxvit'
    families[prefix].append(n)

for fam, mods in sorted(families.items()):
    print(f'  {fam:20s}: {mods}')
````

````python
# ── BackboneOutput Interface ──────────────────────────────────────────────────
import sys
sys.path.insert(0, '../src')
import torch
from hale_core.nn.vision import build_vision_backbone

# Show the unified output interface on a small model
bb = build_vision_backbone('resnet50', img_size=224)
bb.eval()

x = torch.randn(1, 3, 224, 224)
with torch.no_grad():
    out = bb(x)

print('BackboneOutput fields:')
print(f'  pooled_output : {tuple(out.pooled_output.shape)}  ← global embedding [B, D]')
if out.patch_tokens is not None:
    print(f'  patch_tokens  : {tuple(out.patch_tokens.shape)}  ← per-patch tokens [B, N, D]')
else:
    print(f'  patch_tokens  : None  (CNN — use feature_maps for dense tasks)')
print(f'  feature_maps  : {len(out.feature_maps)} scale(s)')
for i, fm in enumerate(out.feature_maps):
    print(f'    [{i}]: {tuple(fm.shape)}')

print()
print('Utility methods:')
print(f'  bb.num_parameters()          = {bb.num_parameters():,}')
print(f'  bb.num_parameters(trainable) = {bb.num_parameters(trainable_only=True):,}')
print(f'  bb.embed_dim                 = {bb.embed_dim}')
print(f'  bb.feature_dims              = {bb.feature_dims}')
````

## Using Backbones in Each Hale* Repo

### Halos-lift (3D lifting / reconstruction)
```python
from lifting.encoders import build_hale_backbone_encoder

encoder = build_hale_backbone_encoder(
    backbone_name='convnext_base',
    img_size=224,
    out_channels=128,
    num_levels=3,          # FPN pyramid levels
    freeze_encoder=False,
)
features = encoder.extract(images)  # list of [B, 128, Hi, Wi]
```

### Hale-VLM (Vision-Language Model)
```python
# In config (vision.yaml):
vision:
  encoder: local            # use HaleBlocks backbone instead of CLIP/SigLIP
  local_backbone:
    backbone_name: dino_vit_base
    freeze_encoder: true
    return_patch_tokens: true

# In code:
from hale_vlm.vision.local_encoders import build_local_vision_tower
tower = build_local_vision_tower(backbone_name='dino_vit_base', ...)
patch_tokens = tower(pixel_values)  # [B, N, D] fed to LLM cross-attention
```

### HALE-WAM (Vision-Language-Action model)
```python
from Halo_VLA.models.vision_backbone import build_halo_vision_backbone

vb = build_halo_vision_backbone(
    backbone_name='swinv2_base',
    img_size=224,
    proj_dim=512,           # optional linear projection
)
# Multi-frame: images [B, T, 3, H, W] → tokens [B, T*N, proj_dim]
tokens = vb(images)
```

## Freezing and Fine-Tuning

```python
bb = build_vision_backbone('vit_base', img_size=224)

# Option 1: Freeze everything (feature extractor only)
bb.freeze_all()

# Option 2: Freeze early stages, fine-tune later stages
bb.freeze_up_to_stage(2)   # stages 0, 1, 2 frozen; stage 3+ trainable

# Option 3: BackboneConfig does all of this declaratively
from hale_core.config.sections.backbone import BackboneConfig
cfg = BackboneConfig(
    name='convnext_base',
    img_size=224,
    freeze_encoder=False,
    freeze_up_to_stage=2,
    pretrained_path='/path/to/weights.safetensors',
)
bb = cfg.build()  # loads weights + applies freeze
```
